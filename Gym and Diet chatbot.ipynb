```python

# ==========================================================
# ADVANCED NLP-BASED SMART GYM & DIET CHATBOT
# ==========================================================

print("===================================================")
print(" NLP-BASED SMART FITNESS CHATBOT 💪🥗")
print("===================================================")
print("Ask about exercises, diet, BMI, calories, recovery")
print("Type 'exit' to stop\n")


# ---------------- HEALTH CALCULATIONS ----------------

def bmi_calculator():
    weight = float(input("Weight (kg): "))
    height = float(input("Height (meters): "))
    bmi = weight / (height ** 2)

    print(f"BMI: {bmi:.2f}")
    if bmi < 18.5:
        print("Category: Underweight")
    elif bmi < 25:
        print("Category: Normal")
    elif bmi < 30:
        print("Category: Overweight")
    else:
        print("Category: Obese")


def calorie_calculator():
    gender = input("Gender (male/female): ").lower()
    age = int(input("Age: "))
    weight = float(input("Weight (kg): "))
    height = float(input("Height (cm): "))

    if gender == "male":
        bmr = 10*weight + 6.25*height - 5*age + 5
    else:
        bmr = 10*weight + 6.25*height - 5*age - 161

    print(f"Daily calories (maintenance): {bmr:.2f} kcal")


# ---------------- MAIN NLP LOOP ----------------

while True:
    user = input("\nYou: ").lower()

    if user == "exit":
        print("Chatbot: Stay consistent. Goodbye 💪")
        break

    # -------- EXERCISES --------
    elif "chest" in user:
        print("Chest exercises: Bench press, incline press, push-ups, cable fly")

    elif "biceps" in user:
        print("Biceps exercises: Barbell curls, dumbbell curls, hammer curls")

    elif "triceps" in user:
        print("Triceps exercises: Dips, skull crushers, pushdowns")

    elif "legs" in user or "leg" in user:
        print("Leg exercises: Squats, lunges, leg press, calf raises")

    elif "back" in user:
        print("Back exercises: Pull-ups, lat pulldown, deadlift, seated row")

    elif "shoulder" in user:
        print("Shoulder exercises: Overhead press, lateral raise, face pulls")

    elif "abs" in user or "core" in user:
        print("Abs exercises: Plank, crunches, leg raises, cable crunch")

    elif "cardio" in user:
        print("Cardio options: Running, cycling, HIIT, skipping")

    # -------- WORKOUT GOALS --------
    elif "weight loss" in user or "fat loss" in user:
        print("For fat loss: Calorie deficit, cardio + strength training")

    elif "muscle gain" in user or "bulking" in user:
        print("For muscle gain: Calorie surplus, progressive overload")

    elif "strength" in user:
        print("Strength training: Low reps, heavy weights, compound lifts")

    # -------- DIET --------
    elif "vegetarian" in user:
        print("Veg protein: Paneer, tofu, dal, chickpeas, soy")

    elif "non veg" in user or "chicken" in user:
        print("Non-veg protein: Chicken, eggs, fish")

    elif "indian diet" in user:
        print("Indian diet: Roti, rice, dal, vegetables, curd")

    elif "post workout" in user:
        print("Post-workout: Protein + carbs (eggs, banana, curd)")

    elif "pre workout" in user:
        print("Pre-workout: Banana, oats, coffee")

    elif "diet" in user:
        print("Balanced diet: Protein, carbs, healthy fats")

    # -------- RECOVERY & SAFETY --------
    elif "rest" in user or "recovery" in user:
        print("Recovery: Sleep 7–8 hours, rest days, stretching")

    elif "injury" in user:
        print("Injury prevention: Warm-up, correct form, avoid ego lifting")

    elif "beginner" in user:
        print("Beginners: Start light, focus on form, be consistent")

    # -------- HEALTH --------
    elif "bmi" in user:
        bmi_calculator()

    elif "calorie" in user:
        calorie_calculator()

    elif "age" in user:
        age = int(input("Enter age: "))
        if age < 18:
            print("Advice: Bodyweight exercises, sports")
        elif age <= 40:
            print("Advice: Strength + cardio")
        else:
            print("Advice: Walking, yoga, light training")

    else:
        print("Chatbot: Ask about exercises, diet, BMI, calories, recovery, or goals.")
```
